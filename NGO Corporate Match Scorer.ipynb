```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

ngos = pd.read_csv("ngo.csv")
companies = pd.read_csv("companies.csv")

# NGO mission (single NGO: Praroop or selected NGO)
ngo_profile = input("Enter NGO mission:")

# Step 1: company text list
company_texts = companies["CSR_Focus"].tolist()

# Step 2: vectorize companies
vectorizer = TfidfVectorizer()
company_vectors = vectorizer.fit_transform(company_texts)

# Step 3: transform NGO into same space
ngo_vector = vectorizer.transform([ngo_profile])

# Step 4: similarity
scores = cosine_similarity(ngo_vector, company_vectors).flatten()

companies["Match_Score"] = scores

# Step 5: rank companies
result = companies.sort_values("Match_Score", ascending=False)

print(result[["Company", "Match_Score"]].head(5))
```
